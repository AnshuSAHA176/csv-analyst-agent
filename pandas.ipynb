```python
import pandas as pd
def csv():
    df = pd.read_csv("demo_ecommerce_orders.csv")
    if df.empty:
        return
    # 1. Total rows and columns
    print(df.shape)

    # 2. Column names and types
    print(df.dtypes)

    # 3. Missing values per column
    print(df.isnull().sum())

    # 4. Duplicate rows
    print(df.duplicated().sum())

    # 5. Basic statistics
    print(df.describe(include="all"))

csv()
```

```python
import pandas as pd

def profile_csv(file_path):
    try:
        df = pd.read_csv(file_path)

        if df.empty:
            raise ValueError("CSV file is empty")

        row, col = df.shape

        return {
            "row_count": int(row),
            "column_count": int(col),
            "duplicate_rows": int(df.duplicated().sum()),
            "missing_values": df.isnull().sum().to_dict(),
            "column_details": df.dtypes.astype(str).to_dict()
        }

    except (pd.errors.EmptyDataError, pd.errors.ParserError) as e:
        raise ValueError(f"Invalid CSV file: {e}") from e

print(profile_csv("demo_ecommerce_orders.csv"))
```
